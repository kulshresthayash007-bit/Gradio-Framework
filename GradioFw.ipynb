```python
import gradio as gr

# 1. Define the backend logic function
def greet(name):
    return f"Hello {name}!"

# 2. Create the interface
demo = gr.Interface(
    fn=greet, 
    inputs="text", 
    outputs="text"
)

# 3. Launch the app inline
demo.launch()
```
